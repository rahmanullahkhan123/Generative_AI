```python
import gradio as gr
from transformers import pipeline
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="google/flan-t5-base"
)

result = generator("Translate English to French: Hello")
print(result)
```

```python
def generate_question(job):

    prompt = f"""
Generate one professional interview question for a {job}.
"""

    result = generator(
        prompt,
        max_new_tokens=60
    )

    return result[0]["generated_text"]
```

```python
generate_question("Machine Learning Engineer")
```

```python
def evaluate_answer(question, answer):

    prompt = f"""
You are an interview coach.

Question:
{question}

Candidate Answer:
{answer}

Evaluate the answer.

Provide:

Score out of 10

Strengths

Weaknesses

Suggestions
"""

    result = generator(
        prompt,
        max_new_tokens=200
    )

    return result[0]["generated_text"]
```

```python
question = "What is Machine Learning?"

answer = """
Machine Learning is a field of AI that enables computers to learn patterns from data without being explicitly programmed.
"""

print(evaluate_answer(question, answer))
```

```python
def follow_up(question, answer):

    prompt = f"""
Question:
{question}

Candidate Answer:
{answer}

Generate one follow-up interview question.
"""

    result = generator(
        prompt,
        max_new_tokens=80
    )

    return result[0]["generated_text"]
```

```python
def interview(job, answer):

    question = generate_question(job)

    feedback = evaluate_answer(question, answer)

    next_question = follow_up(question, answer)

    return question, feedback, next_question
```

```python
demo = gr.Interface(
    fn=interview,

    inputs=[
        gr.Textbox(label="Job Role"),
        gr.Textbox(
            label="Your Answer",
            lines=6
        )
    ],

    outputs=[
        gr.Textbox(label="Interview Question"),
        gr.Textbox(label="AI Feedback"),
        gr.Textbox(label="Follow-up Question")
    ],

    title="AI Interview Preparation Coach",

    description="Practice interviews using AI."
)

demo.launch()
```
